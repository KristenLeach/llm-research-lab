## The 5-Item Reproducibility Checklist

Before sharing any research analysis, verify all five of these:

| # | Item | What it means |
|---|---|---|
| 1 | **Seeds set** | Any random number generators are seeded with a fixed value (`random.seed(42)`, `np.random.seed(42)`) |
| 2 | **Data path documented** | The exact path or URL of every input file is recorded somewhere findable |
| 3 | **Output saved** | Results are written to a file, not just printed |
| 4 | **Dependencies pinned** | The exact package versions used are recorded (e.g., `requirements.txt` or `pyproject.toml`) |
| 5 | **README written** | A plain-language file explains what the analysis does and how to re-run it |

Failing any one of these can make results impossible to reproduce:
- No seed → random results change on every run
- No data path → nobody knows where the data came from
- No saved output → the numbers exist only in memory
- No pinned deps → a package update silently changes behaviour
- No README → even the original author forgets how to run it

## Wrap-Up

The reproducibility checklist is a simple but powerful habit:

1. **Set seeds** before any random operation
2. **Document data paths** explicitly — no hard-coded absolute paths
3. **Save outputs** to files, not just memory
4. **Pin dependencies** with `pyproject.toml` or `requirements.txt`
5. **Write a README** before sharing or publishing

The `reproducibility_report` dictionary you built in Exercise 3 is a lightweight version of what research labs sometimes call a **model card** or **experiment log** — structured metadata that travels with the results.
