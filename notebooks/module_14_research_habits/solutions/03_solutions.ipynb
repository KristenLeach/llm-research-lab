## Exercise 1 — Solution

A list comprehension over `checklist.items()` collects keys whose value is `False`. Since only `readme_written` is `False`, the resulting list has exactly one item.

```
checklist = {
    'seeds_set': True,
    'data_path_documented': True,
    'output_saved': True,
    'deps_pinned': True,
    'readme_written': False
}

incomplete = [k for k, v in checklist.items() if not v]
```

```
check_type(incomplete, list, "incomplete is a list")
check_contains(incomplete, 'readme_written', "incomplete includes 'readme_written'")
check_length(incomplete, 1, "exactly one item is incomplete")
```
