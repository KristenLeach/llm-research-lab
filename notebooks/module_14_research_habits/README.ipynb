# Module 14 — Research-Style Coding Habits

Code quality matters differently in research than in production software.
Research code needs to be readable (so colleagues can reproduce and verify it),
reproducible (so results hold up), and well-documented (so findings can be trusted).
This module covers the habits that distinguish good research code from messy notebooks.

---

## Learning Objectives

By the end of this module, you will be able to:

- Know when to use a notebook vs. a Python script
- Apply version control basics with git to research code and data analysis
- Use consistent naming conventions for variables, files, and functions
- Write docstrings and inline comments that explain *why*, not just *what*
- Apply a code review mindset to your own work (read it as a stranger would)
- Follow a reproducibility checklist before sharing results

---

## What You'll Build

**Mini-project:** Refactor a messy analysis notebook into clean, documented, reproducible research code:
- Start with a provided "messy" notebook (magic numbers, no docstrings, inconsistent style)
- Extract reusable functions into a `.py` module
- Add docstrings and type hints
- Add assertions and input validation
- Set random seeds
- Write a short README explaining how to reproduce the analysis

## Key Concepts

- **Notebook vs. script**: notebooks for exploration and communication; scripts for production pipelines
- **git basics**: `commit`, `diff`, `log` — track changes to code over time
- **Naming conventions**: `snake_case` for Python; descriptive, not abbreviated names
- **Docstring**: a string at the top of a function explaining its purpose, args, and return value
- **Type hints**: `def compute(scores: list[float]) -> float:` — optional but helpful for clarity
- **Magic numbers**: unexplained numeric constants — replace with named variables
- **Reproducibility checklist**: seeds set, dependencies pinned, data paths documented, output saved
- **Code review**: reading your own code as if someone else wrote it
