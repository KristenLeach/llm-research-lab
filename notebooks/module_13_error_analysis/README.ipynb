# Module 13 — Error Analysis

After you've built and evaluated a classifier, the next step is understanding
*why* it fails. Error analysis is the practice of systematically examining
mistakes to find patterns, prioritize improvements, and surface safety-relevant failure modes.

---

## Learning Objectives

By the end of this module, you will be able to:

- Build and interpret a confusion matrix
- Analyze false positives and false negatives qualitatively (look at the actual examples)
- Slice error rates by category, model version, or other metadata
- Identify systematic failure modes (patterns in the errors)
- Propose concrete improvements based on your error analysis
- Write up an error analysis section for a research report

---

## What You'll Build

**Mini-project:** Full error analysis on a safety classifier:
- Build a confusion matrix and visualize it as a heatmap
- Extract all false positives and false negatives
- Look at 5-10 examples from each error category — what do they have in common?
- Slice error rates by task category — is one category harder than others?
- Propose 2-3 specific improvements to reduce the most damaging errors

## Key Concepts

- **Confusion matrix**: a 2×2 (or N×N) table showing true/false positives/negatives
- **True positive (TP)**: correctly flagged as unsafe
- **True negative (TN)**: correctly identified as safe
- **False positive (FP)**: incorrectly flagged as unsafe (safe output, wrongly flagged)
- **False negative (FN)**: incorrectly identified as safe (unsafe output, missed)
- **Slice analysis**: computing metrics for subgroups to find where errors concentrate
- **Systematic failure**: when errors are not random but share a common cause
- **Qualitative review**: actually reading the failure cases, not just computing numbers
