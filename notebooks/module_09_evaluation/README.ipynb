# Module 09 — Evaluation Methodology

How do you actually know if a model is better? Evaluation methodology is the practice of
designing rigorous tests that produce trustworthy answers to that question.
Bad evaluation design leads to misleading conclusions — a critical concern in AI safety.

---

## Learning Objectives

By the end of this module, you will be able to:

- Define clear evaluation tasks and the behaviors you're trying to measure
- Choose appropriate metrics for different types of model outputs
- Build a baseline to compare against (random, majority class, prior version)
- Identify and avoid data contamination (train/test leakage)
- Measure inter-rater agreement (Cohen's kappa) when humans are judging outputs
- Structure a small evaluation pipeline end to end

---

## What You'll Build

**Mini-project:** Design and run a small evaluation of model honesty across 3 task types:
- Define a rubric for what counts as "honest" in each task type
- Create a test set of 10-20 prompts per task type
- Define your metric (e.g., fraction of responses rated as honest)
- Implement a baseline (e.g., random classifier, most-common-rating classifier)
- Write up the methodology as if reporting it to a colleague

## Key Concepts

- **Evaluation task**: a specific, well-defined prompt or scenario for testing the model
- **Metric**: a number that summarizes performance (accuracy, F1, win rate, etc.)
- **Baseline**: the simplest comparison point — if you can't beat it, your model is not useful
- **Hold-out set**: data reserved for final evaluation, never used in development
- **Data contamination**: when test data leaks into training, making results falsely optimistic
- **Inter-rater agreement**: how consistently multiple human raters agree on the same outputs
- **Cohen's kappa**: a metric for inter-rater agreement that accounts for chance
- **Construct validity**: does your metric actually measure what you think it measures?
