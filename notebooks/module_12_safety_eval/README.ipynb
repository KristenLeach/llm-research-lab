# Module 12 — Safety-Oriented Classification & Evaluation

AI safety research often involves classifying model outputs as safe or unsafe,
helpful or harmful, honest or deceptive. This module covers how to build, tune,
and evaluate those classifiers — and how to think carefully about the tradeoffs.

---

## Learning Objectives

By the end of this module, you will be able to:

- Build rule-based classifiers for model output safety (keyword matching, heuristics)
- Tune classification thresholds and understand the precision/recall tradeoff
- Compute precision, recall, F1, and accuracy for a binary classifier
- Understand the asymmetric cost of false positives vs. false negatives in safety contexts
- Evaluate a classifier against a labeled dataset
- Think through what failure modes matter most for safety-critical classification

## Key Concepts

- **Binary classification**: categorizing each output as one of two classes (safe/unsafe)
- **Precision**: of the things we flagged, what fraction were actually unsafe?
- **Recall**: of all the actually unsafe things, what fraction did we catch?
- **F1 score**: the harmonic mean of precision and recall
- **False positive**: flagging something that's actually safe (over-cautious)
- **False negative**: missing something that's actually unsafe (under-cautious)
- **Threshold**: the score above which we classify as "unsafe" — tuning it moves the precision/recall tradeoff
- **Asymmetric costs**: in safety contexts, false negatives are often much more costly than false positives
