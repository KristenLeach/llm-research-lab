## Exercise 1 Solution

```python
raw = [0.95, 0.98, 0.91, 0.96, 0.83]

a2_scores = np.array(raw)
# float() converts numpy scalar → Python float, which check_approx expects
a2_mean   = round(float(np.mean(a2_scores)), 3)
a2_max    = float(np.max(a2_scores))
```

```python
check_type(a2_scores, np.ndarray, "a2_scores is an ndarray")
check_length(a2_scores, 5, "a2_scores has 5 elements")
check_approx(a2_mean, 0.926, tolerance=1e-3, label="a2_mean")
check_approx(a2_max, 0.98, tolerance=1e-6, label="a2_max")
```

## Exercise 2 Solution

```python
scores = np.array([0.92, 0.96, 0.88, 0.94, 0.78, 0.71, 0.60, 0.65, 0.82, 0.74,
                   0.95, 0.98, 0.91, 0.96, 0.83, 0.78, 0.72, 0.70, 0.85, 0.77])

passing_mask   = scores > 0.8
passing_scores = scores[passing_mask]
# int() because np.sum returns a numpy integer; check_equal compares with Python int
n_passing      = int(np.sum(passing_mask))

# Common mistake: using >= instead of > would give 12 (includes 0.82 and 0.85)
```

```python
check_type(passing_mask, np.ndarray, "passing_mask is an ndarray")
check_type(passing_scores, np.ndarray, "passing_scores is an ndarray")
check_equal(int(n_passing), 11, "11 scores above 0.8")
check_length(passing_scores, 11, "passing_scores has 11 elements")
```

## Exercise 3 Solution

```python
a2 = np.array([0.95, 0.98, 0.91, 0.96, 0.83])
b1 = np.array([0.71, 0.60, 0.65, 0.82, 0.74])

improvement      = a2 - b1
mean_improvement = round(float(np.mean(improvement)), 3)
best_task_idx    = int(np.argmax(improvement))
# Index 1 = harmful_refusal — the task with the biggest safety gap (+0.38)
# This is the most practically significant improvement between the two models
```

```python
check_type(improvement, np.ndarray, "improvement is an ndarray")
check_approx(mean_improvement, 0.222, tolerance=1e-3, label="mean_improvement")
check_equal(int(best_task_idx), 1, "largest improvement at index 1 (harmful_refusal)")
```
