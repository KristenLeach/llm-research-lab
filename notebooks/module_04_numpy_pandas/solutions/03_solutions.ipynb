# Module 04 — Notebook 3 Solutions: groupby and Aggregation

```python
import sys
sys.path.insert(0, "../../../")
from src.checks import check_equal, check_type, check_approx, check_length, check_contains
import numpy as np
import pandas as pd
from pathlib import Path

DATA_PATH = Path("../../../data/synthetic/evaluation_results.csv")
df = pd.read_csv(DATA_PATH)
```

## Exercise 1 Solution

```python
per_model_mean = df.groupby("model")["score"].mean()
best_model     = per_model_mean.idxmax()   # returns the index label, not the value
b1_mean        = round(float(per_model_mean["model-b-v1"]), 3)
```

```python
check_type(per_model_mean, pd.Series, "per_model_mean is a Series")
check_equal(best_model, "model-a-v2", "best model is model-a-v2")
check_approx(b1_mean, 0.704, tolerance=1e-3, label="b1_mean")
```

## Exercise 2 Solution

```python
task_summary = df.groupby("task")["score"].agg(["mean", "min", "max", "std"])
# .loc[row_label, col_label] accesses a specific cell by both labels
refusal_mean = round(float(task_summary.loc["harmful_refusal", "mean"]), 3)
```

```python
check_type(task_summary, pd.DataFrame, "task_summary is a DataFrame")
check_contains(list(task_summary.columns), "mean", "task_summary has mean column")
check_approx(refusal_mean, 0.815, tolerance=1e-3, label="refusal_mean")
```

## Exercise 3 Solution

```python
df = pd.read_csv(DATA_PATH)

df["pass"] = df["score"] >= 0.85

# True sums to 1, False to 0 — .sum() on a boolean column counts Trues
passing_per_model = df.groupby("model")["pass"].sum()
a2_passing        = int(passing_per_model["model-a-v2"])

# model-a-v2 scores: 0.95✓ 0.98✓ 0.91✓ 0.96✓ 0.83✗ → 4
# model-b-v1 scores: 0.71✗ 0.60✗ 0.65✗ 0.82✗ 0.74✗ → 0
```

```python
check_type(passing_per_model, pd.Series, "passing_per_model is a Series")
check_equal(int(a2_passing), 4, "model-a-v2 has 4 tasks passing threshold 0.85")
check_equal(int(passing_per_model["model-b-v1"]), 0, "model-b-v1 has 0 tasks passing threshold 0.85")
```
