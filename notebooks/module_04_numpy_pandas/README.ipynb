# Module 04 — NumPy & Pandas Basics

NumPy and pandas are the foundation of almost all Python data analysis.
NumPy gives you fast arrays; pandas gives you labeled tables (DataFrames).
If you're coming from JavaScript, think of a DataFrame as a typed, queryable array of objects
with built-in statistics and grouping operations.

---

## Key Concepts

- **ndarray**: NumPy's n-dimensional array type
- **Vectorized operations**: apply math to an entire array at once, no loop needed
- **Broadcasting**: rules for combining arrays of different shapes
- **DataFrame**: pandas' 2D labeled data structure (like a spreadsheet)
- **Series**: a single column of a DataFrame
- **Boolean indexing**: `df[df["col"] > value]` — filter rows by condition
- **groupby / agg**: split data into groups, compute statistics per group
- **NaN**: "not a number" — how pandas represents missing data
