## 1. The 5-Section Memo Structure

A good research memo has five sections. Each section has a specific job:

| Section | Purpose | Typical length |
|---------|---------|----------------|
| Question | What are you trying to answer? | 1-2 sentences |
| Method | How did you approach it? | 2-4 sentences |
| Results | What did you find? | 3-6 sentences with numbers |
| Caveats | What could be wrong or misleading? | 2-4 sentences |
| Next Steps | What should happen based on this? | 2-4 bullet points |

This structure keeps memos predictable. A reader can jump straight to **Results** to get the finding, or to **Caveats** to check for limitations — without reading linearly from start to finish.

## 2. The Lede

"Burying the lede" is a journalism term for hiding the most important point deep in the story. In a research memo, the equivalent is putting your key finding in the last paragraph.

**Don't do this.** Lead with the most important thing.

Newspaper analogy: front-page headlines don't say "We looked at the city council. Here is some background. After some analysis... the mayor resigned." They say: **Mayor Resigns.** Then the details follow.

In a research memo:
- **Bad lede**: "We loaded model_outputs.json and ran two classifiers..."
- **Good lede**: "Classifier v2 achieves F1=0.86, a major improvement over the 0.0 baseline."

Start with the finding. Earn the reader's attention.
