## 1. Types of Caveats

A caveat is a limitation that could make your conclusion wrong or less general. Common types:

- **Small dataset** — results on 20 examples may not hold on 2,000
- **Domain limitation** — the classifier was built for this dataset; it may not generalize
- **Simplified model** — keyword matching misses many failure modes (e.g., factual errors, subtle bias)
- **Can't generalize** — both models in the dataset are from the same family; other model families may behave differently

The rule: **flag what could make your conclusion wrong, but don't undersell real findings.**

Bad caveat (undersells): "These results are preliminary and should not be trusted."
Good caveat: "Results are based on 20 examples; precision/recall estimates may shift by ±0.1 on a larger set."

## 2. Module 12 Caveats — Worked Example

Here are three honest caveats for the Module 12 analysis:

1. **Small dataset**: 20 examples is too few to reliably estimate precision and recall. A change of one example flips F1 by ~0.05.

## 3. Specific vs. Vague Next Steps

Next steps should be specific enough that someone could act on them without asking a follow-up question.

| Vague | Specific |
|-------|----------|
| "Improve the classifier" | "Add 'visible from space' to TRIGGERS to catch out_011" |
| "Do more testing" | "Evaluate on a held-out set of 50 new examples" |

A good next step names:
- **What** to do (concrete action)
- **How much** or **what scope** (e.g., 50 examples, 3 new trigger words)
- Optionally: **why** it is the priority (which failure mode it addresses)

Vague next steps feel like they cover everything, but they authorize nothing.
