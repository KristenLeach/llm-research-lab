# Module 15 — Writing Research Memos from Findings

Producing good numbers is only half the job. The other half is communicating what they mean.
A research memo is a short, structured document that presents findings clearly enough
for a colleague to understand, trust, and act on them.

This module bridges technical analysis and written communication — a skill that distinguishes
good researchers from great ones.

---

## Learning Objectives

By the end of this module, you will be able to:

- Structure a research memo: question, method, results, caveats, next steps
- Present quantitative results clearly: numbers in context, not just raw values
- Write appropriate caveats without underselling your findings
- Suggest concrete, prioritized next steps based on your findings
- Write the same findings for two different audiences: a technical peer and a non-technical stakeholder
- Avoid common memo mistakes: burying the lede, overclaiming, ignoring limitations

## What You'll Build

**Mini-project:** Write a 1-page research memo from the Module 12 safety evaluation results:
- **Question**: What is the performance of our rule-based safety classifier on model_outputs.json?

- **Results**: Key numbers (precision, recall, F1) with context and comparison to baseline
- **Caveats**: What the numbers don't capture, what could be misleading
- **Next Steps**: 2-3 specific, prioritized follow-up actions

## Key Concepts

- **Research memo**: a short (1-3 page) internal document presenting specific findings
- **Lede**: the most important finding — put it first, not last
- **Numbers in context**: "F1 = 0.72" means little; "F1 = 0.72, vs. 0.61 for a random baseline" is informative
- **Caveats**: honest acknowledgment of what could be wrong with your analysis
- **Overclaiming**: stating stronger conclusions than your data supports
- **Audience calibration**: adjusting vocabulary and depth based on who will read the memo
- **Next steps**: actions that follow logically from your findings — specific, not vague
