## Why This Matters for AI Research Engineering

Raw numbers are meaningless without context. "F1 = 0.86" — is that good? Bad? Nobody knows without a reference point.

At AI labs, every number you report should be paired with a baseline, comparison, or benchmark. A finding like "our classifier improved" is not a finding at all. "Our classifier improved from F1=0.83 to F1=0.86, compared to a majority-class baseline of F1=0.0" — that is a finding.

## 1. Three Types of Context

When you report a number, give it one of these three anchors:

**1. Baseline comparison** — How does it compare to doing nothing (or doing the simplest possible thing)?

> "F1=0.86 vs. F1=0.0 for a majority-class baseline"

**2. Version comparison** — How does it compare to the previous approach?

> "F1=0.86 for v2 vs. F1=0.83 for v1"

**3. Absolute benchmark** — How does it compare to a known reference point (e.g., human performance, published benchmarks)?

> "F1=0.86 vs. ~0.92 estimated human agreement on this task"

For Module 12, we have types 1 and 2. That is enough to make the numbers meaningful.

## 2. Worked Example

Here is how context transforms a number:

**Without context:**
> F1 = 0.86

Is that good? We have no idea.

**With baseline comparison:**
> F1 = 0.86 for classifier v2, compared to 0.0 for a majority-class baseline.

Now we know: the classifier is dramatically better than guessing.

**With version comparison added:**
> F1 = 0.86 for classifier v2, up from 0.83 for v1, and compared to 0.0 for a majority-class baseline.

Now we see both that there was improvement and that both versions substantially beat the baseline.

Practical rule: **never report a lone number.** Always add at least one anchor.
