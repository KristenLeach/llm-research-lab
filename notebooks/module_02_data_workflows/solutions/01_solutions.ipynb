## Exercise 1 Solution: Using `math.sqrt`

**Task:** Use `math.sqrt()` to compute the square root of 144.

```python
import math

result = math.sqrt(144)
print(result)  # 12.0

# Note: math.sqrt() always returns a float, even when the result is a whole number.
# If you need an integer, you can cast it: int(math.sqrt(144))
print(type(result))  # <class 'float'>
```

## Exercise 2 Solution: `pathlib` Path Object

**Task:** Use `pathlib.Path` to create a path object pointing to a data file,
and demonstrate how to work with it.

```python
from pathlib import Path

# Create a path object for a data file
data_file = Path("../../../data/synthetic/model_outputs.json")

# pathlib gives you handy properties — no more string splitting!
print("Full path:", data_file)
print("File name:", data_file.name)          # model_outputs.json
print("Stem (no extension):", data_file.stem) # model_outputs
print("Extension:", data_file.suffix)         # .json
print("Parent directory:", data_file.parent)  # ../../../data/synthetic

# Check if the file exists
print("File exists:", data_file.exists())

# Build paths by joining with /  (much cleaner than os.path.join!)
data_dir = Path("../../../data/synthetic")
csv_file = data_dir / "evaluation_results.csv"
print("CSV path:", csv_file)
```

## Exercise 3 Solution: `random.seed` and `random.choice`

**Task:** Set a random seed with `random.seed(42)`, then demonstrate `random.choice`.
Setting a seed ensures reproducible results — critical in research.

```python
import random

# Setting a seed makes random choices reproducible.
# Anyone who runs this code will get the same sequence.
random.seed(42)

models = ["gpt-4", "claude-3", "gemini-pro", "llama-3"]

# random.choice picks one element at random
chosen = random.choice(models)
print("Chosen model:", chosen)  # Will always be the same with seed=42

# random.choices (plural) picks multiple with replacement
sample = random.choices(models, k=3)
print("Sample of 3:", sample)

# random.sample picks multiple WITHOUT replacement
unique_sample = random.sample(models, k=2)
print("Unique sample of 2:", unique_sample)

# Re-run the seed to confirm reproducibility
random.seed(42)
chosen_again = random.choice(models)
print("Chosen again (same seed):", chosen_again)  # Same as 'chosen' above
print("Results match:", chosen == chosen_again)
```

## Key Takeaways

- `math` gives you standard mathematical functions — use `math.sqrt()`, `math.pi`, `math.log()`, etc.
- `pathlib.Path` is the modern way to handle file paths — prefer it over string manipulation or `os.path`
- `random.seed()` makes random operations reproducible — always set a seed in research code
- Use `from module import name` to import specific items; use `import module` to import the whole module
