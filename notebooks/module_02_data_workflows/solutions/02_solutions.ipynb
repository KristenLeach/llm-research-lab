## Exercise 1 Solution: Write Model Names to a File, Then Read Them Back

**Task:** Write a list of model names to a text file, one per line, then read the file back and print each name.

```python
import os

# Always create the output directory if it doesn't exist.
# exist_ok=True means no error if the directory is already there.
os.makedirs("output", exist_ok=True)

model_names = ["gpt-4", "claude-3-opus", "gemini-pro", "llama-3-70b"]

# --- WRITE ---
# Use 'with' so the file is automatically closed when the block exits.
# Mode 'w' means write (creates the file, or overwrites if it already exists).
filepath = "output/models.txt"

with open(filepath, "w") as f:
    for name in model_names:
        f.write(name + "\n")  # Write each name followed by a newline

print(f"Wrote {len(model_names)} model names to {filepath}")

# --- READ ---
# Mode 'r' means read (the default — you can omit it, but explicit is clearer).
with open(filepath, "r") as f:
    contents = f.read()

print("\nFile contents:")
print(contents)

# You can also read line by line:
print("Reading line by line:")
with open(filepath, "r") as f:
    for line in f:
        print(" -", line.strip())  # .strip() removes the trailing newline
```

## Exercise 2 Solution: Count Lines in the File

**Task:** Read the file you wrote and count how many lines it has.

```python
filepath = "output/models.txt"

# Method 1: readlines() returns a list — one element per line
with open(filepath, "r") as f:
    lines = f.readlines()

print(f"Number of lines (readlines): {len(lines)}")

# Method 2: iterate and count (memory-efficient for large files)
count = 0
with open(filepath, "r") as f:
    for line in f:
        if line.strip():  # Only count non-empty lines
            count += 1

print(f"Number of non-empty lines: {count}")

# For large files, method 2 is better — it doesn't load everything into memory at once.
# For small files like this, either works fine.
```

## Exercise 3 Solution: Append a New Model Name

**Task:** Append a new model name to the existing file without overwriting the other names.

```python
import os

os.makedirs("output", exist_ok=True)
filepath = "output/models.txt"

new_model = "mistral-large"

# Mode 'a' means APPEND — adds to the end of the file without erasing existing content.
# This is different from 'w' which would wipe the file!
with open(filepath, "a") as f:
    f.write(new_model + "\n")

print(f"Appended '{new_model}' to {filepath}")

# Verify by reading the whole file
print("\nUpdated file contents:")
with open(filepath, "r") as f:
    for line in f:
        print(" -", line.strip())

# Count lines to confirm we now have 5
with open(filepath, "r") as f:
    line_count = sum(1 for line in f if line.strip())
print(f"\nTotal models: {line_count}")
```

---
## Key Takeaways

- Always use `with open(...)` — it guarantees the file is closed even if an error occurs
- File modes: `'r'` = read, `'w'` = write (overwrites!), `'a'` = append, `'rb'`/`'wb'` = binary
- `os.makedirs(path, exist_ok=True)` before writing — prevents errors if the directory doesn't exist
- `.strip()` removes leading/trailing whitespace including `\n` — use it when reading lines
- For large files, iterate line by line instead of loading everything with `.read()` or `.readlines()`
