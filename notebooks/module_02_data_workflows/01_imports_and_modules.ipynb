## What is a module?

A **module** is just a Python file (`.py`) that contains code — functions, classes, variables — that you can reuse in other files.

If you're coming from JavaScript, modules map directly to what you know:

| JavaScript | Python |
|---|---|
| `const fs = require('fs')` | `import os` |
| `const { readFile } = require('fs')` | `from os.path import join` |
| `import { sqrt } from 'mathlib'` | `from math import sqrt` |
| `npm install numpy` | `pip install numpy` |

Python ships with a large **standard library** of built-in modules (no install needed). You'll also install third-party packages like `numpy`, `pandas`, and `transformers` from PyPI using `pip`.

## `import math` — two import styles

There are two common ways to import:

1. **`import math`** — imports the whole module; access names with `math.sqrt()`
2. **`from math import sqrt`** — imports a specific name; call `sqrt()` directly

Both are correct. Use style 1 when the module name adds clarity (`math.sqrt` is clear). Use style 2 when you're using something constantly and the module name is noise.

````python
import math

# Access via the module name
print(math.sqrt(16))    # 4.0
print(math.pi)          # 3.141592653589793
print(math.log(100, 10))  # 2.0  (log base 10 of 100)

# Import a specific name directly
from math import sqrt, pi

print(sqrt(16))  # 4.0 — no "math." prefix needed
print(pi)        # 3.141592653589793
````

````python
import os

# Current working directory (like process.cwd() in Node)
cwd = os.getcwd()
print("Working directory:", cwd)

# Build a path safely (handles / vs \ on different OSes)
path = os.path.join("data", "synthetic", "model_outputs.json")
print("Joined path:", path)

# List files in a directory
entries = os.listdir(".")
print("Files here:", entries[:5])  # first 5 entries

# Create a directory (won't error if it already exists)
os.makedirs("output", exist_ok=True)
print("output/ directory ready")
````

## `pathlib.Path` — the modern way to handle file paths

`pathlib` was added in Python 3.4 and is now the preferred way to work with paths.
Instead of string manipulation, you get an object with useful properties and methods.

Think of it as a typed wrapper around a file path string.

````python
from pathlib import Path

# Create a Path object — doesn't have to exist yet
p = Path("data/synthetic/model_outputs.json")

print("Path:    ", p)
print(".name:   ", p.name)     # 'model_outputs.json' — filename with extension
print(".stem:   ", p.stem)     # 'model_outputs' — filename without extension
print(".suffix: ", p.suffix)   # '.json' — the extension
print(".parent: ", p.parent)   # 'data/synthetic' — the containing directory
print(".exists():", p.exists())  # True if the file actually exists on disk

# Build paths with / operator (cleaner than os.path.join)
repo_root = Path("../../")
data_path = repo_root / "data" / "synthetic" / "model_outputs.json"
print("\nBuilt path:", data_path)
````

## `import random` — randomness and reproducibility

The `random` module generates random numbers and makes random choices.
In research, **reproducibility** is critical: use `random.seed()` to lock in a starting point
so that "random" choices are the same every run.

````python
import random

models = ["model-a-v1", "model-b-v1", "model-a-v2", "model-b-v2"]

# Without a seed, results change every run
print("Random choice:", random.choice(models))

# With a seed, results are reproducible — same seed = same sequence
random.seed(42)
print("Seeded choice:", random.choice(models))  # Always the same with seed=42

# shuffle modifies the list in place (like Array.prototype.sort with random comparator)
sample = ["a", "b", "c", "d", "e"]
random.seed(42)
random.shuffle(sample)
print("Shuffled:", sample)  # Always the same shuffle with seed=42

# random.sample — pick k unique items without replacement
random.seed(42)
picked = random.sample(models, k=2)
print("Sampled 2:", picked)
````

## Preview: `if __name__ == "__main__"`

You'll often see this pattern at the bottom of Python scripts:

```python
if __name__ == "__main__":
    main()
```

**What it means:** When Python runs a file directly (e.g., `python my_script.py`), it sets the special variable `__name__` to the string `"__main__"`. When the file is *imported* as a module by another file, `__name__` is set to the module's name instead.

This guard prevents "setup" code (like running a script) from firing when the file is just being imported for its functions. You'll see this everywhere in Python projects. We'll cover it in depth in Module 06.

## Why This Matters for AI Research Engineering

In research, you import tools constantly:

- `numpy` for numerical arrays and linear algebra
- `pandas` for loading and summarizing tabular data
- `pathlib` for navigating file paths without hardcoding separators
- `json` for reading model outputs from APIs
- Third-party packages like `transformers`, `torch`, `openai`

Understanding how imports work — the difference between `import X` and `from X import Y`,
how `sys.path` controls where Python looks, and what `__name__` does — prevents confusing
errors and lets you find and use the right tools confidently.

Reproducibility (`random.seed`) is especially important in AI safety research: if you can't
