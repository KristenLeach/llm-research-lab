# Module 05 — Plotting & Exploratory Analysis

A chart communicates results faster than a table of numbers.
This module covers matplotlib and seaborn — the two most common Python plotting libraries —
and teaches you how to create the visualizations you'll use most often in AI research.

---

By the end of this module, you will be able to:

- Create bar charts to compare model performance across tasks

- Customize plots: titles, axis labels, color palettes, figure size

- Save plots to files for research reports and memos
- Know when to use which chart type

## Key Concepts

- **matplotlib**: the foundational Python plotting library — explicit, flexible, verbose
- **seaborn**: built on matplotlib, higher-level, better defaults for statistical plots
- **Figure and Axes**: matplotlib's object model — a Figure contains one or more Axes (subplots)
- **`plt.subplots()`**: create a figure with one or more subplot panels
- **`ax.set_title()`, `ax.set_xlabel()`**: customize axis labels
- **`plt.savefig()`**: save a plot to a PNG or PDF file
- **Exploratory vs. presentation plots**: quick-and-dirty for exploration, polished for sharing
