# Module 05 — Notebook 1 Solutions: matplotlib Basics

```python
import sys
sys.path.insert(0, "../../../")
from src.checks import check_equal, check_type, check_approx, check_length
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib
from pathlib import Path
%matplotlib inline
DATA_PATH = Path("../../../data/synthetic/evaluation_results.csv")
OUTPUT_DIR = Path("../../../output/plots")
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
df = pd.read_csv(DATA_PATH)
```

## Exercise 1 Solution

```python
per_task_means = df.groupby("task")["score"].mean().sort_values()

fig, ax = plt.subplots(figsize=(9, 5))
ax.barh(per_task_means.index, per_task_means.values, color="steelblue", alpha=0.85)
ax.axvline(per_task_means.mean(), color="crimson", linestyle="--",
           label=f"Mean ({per_task_means.mean():.3f})")
ax.set_title("Mean Score by Task")
ax.set_xlabel("Mean Score")
ax.set_xlim(0, 1.05)
ax.legend()
plt.tight_layout()
plt.show()
```

```python
check_type(per_task_means, pd.Series, "per_task_means is a Series")
check_length(per_task_means, 5, "per_task_means has 5 tasks")
check_approx(float(per_task_means["instruction_following"]), 0.8925, 1e-3, "instruction_following mean")
check_equal(isinstance(fig, plt.Figure), True, "fig is a matplotlib Figure")
check_equal(isinstance(ax, matplotlib.axes.Axes), True, "ax is a matplotlib Axes")
```

## Exercise 2 Solution

```python
a_df = df[df["model"].isin(["model-a-v1", "model-a-v2"])]
# pivot_table with tasks as index, models as columns
a_pivot = a_df.pivot_table(index="task", columns="model", values="score")
# v2 - v1 for harmful_refusal: 0.98 - 0.96 = 0.02
refusal_gain = round(float(
    a_pivot.loc["harmful_refusal", "model-a-v2"] - a_pivot.loc["harmful_refusal", "model-a-v1"]
), 2)

# Grouped bar chart using DataFrame.plot(kind='bar')
a_pivot.plot(kind="bar", figsize=(10, 5), title="model-a v1 vs v2 per task")
plt.tight_layout()
plt.show()
```

```python
check_type(a_df, pd.DataFrame, "a_df is a DataFrame")
check_equal(len(a_df), 10, "a_df has 10 rows")
check_type(a_pivot, pd.DataFrame, "a_pivot is a DataFrame")
check_approx(refusal_gain, 0.02, 1e-2, "refusal_gain")
```

## Exercise 3 Solution

```python
per_model = df.groupby("model")["score"].mean().sort_values(ascending=False)
fig, ax = plt.subplots(figsize=(8, 5))
ax.bar(per_model.index, per_model.values, color="steelblue", alpha=0.85)
ax.set_title("Mean Score by Model")
ax.set_ylabel("Mean Score")
plt.tight_layout()

save_path = Path(OUTPUT_DIR / "my_model_bars.png")
fig.savefig(save_path, dpi=150, bbox_inches="tight")
print(f"Saved: {save_path}")
plt.show()
```

```python
check_type(save_path, Path, "save_path is a Path")
check_equal(save_path.exists(), True, "plot file was saved to disk")
```
