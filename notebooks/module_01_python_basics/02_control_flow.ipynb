## 1. Conditionals: `if`, `elif`, `else`

Python uses **indentation** to define blocks — there are no curly braces. This is the biggest syntactic difference from JavaScript.

```javascript
// JavaScript
if (score >= 0.9) {
  console.log("excellent");
} else if (score >= 0.7) {
  console.log("good");
} else {
  console.log("needs review");
}
```

```python
# Python
if score >= 0.9:
    print("excellent")
elif score >= 0.7:
    print("good")
else:
    print("needs review")
```

Key points:
- Condition ends with a colon `:`
- Body is indented (conventionally 4 spaces)
- `elif` instead of `else if`
- No parentheses required around the condition (though they're allowed)

````python
# Conditional example
score = 0.85

if score >= 0.9:
    label = "excellent"
elif score >= 0.7:
    label = "good"
else:
    label = "needs review"

print(f"Score {score} → {label}")
````

## 2. Comparison and Boolean Operators

Comparison operators work the same as in JavaScript:

| Operator | Meaning |
|----------|---------|
| `==` | equal to |
| `!=` | not equal to |
| `<` | less than |
| `>` | greater than |
| `<=` | less than or equal |
| `>=` | greater than or equal |

Boolean operators use **words**, not symbols:

| Python | JavaScript equivalent |
|--------|----------------------|
| `and` | `&&` |
| `or` | `\|\|` |
| `not` | `!` |

Python does **not** have `===`. The `==` operator compares values, and Python is strict about types (e.g., `1 == True` is `True` because `True` is `1` internally, but `"1" == 1` is `False`).

````python
# Comparison and boolean operators
score = 0.85
threshold = 0.7
model = "gpt-4"

print(score >= threshold)          # True
print(score < 0.9)                 # True
print(score >= threshold and score < 0.9)  # True (score is between 0.7 and 0.9)
print(model == "gpt-4" or model == "claude-3")  # True
print(not (score < threshold))     # True (score is NOT below threshold)

# Python 'in' for membership — often cleaner than OR chains
approved_models = ["gpt-4", "claude-3", "gemini-pro"]
print(model in approved_models)    # True
````

## 3. `for` Loops

Python's `for` loop iterates directly over items in a collection — no index needed unless you want one.

```javascript
// JavaScript
for (const item of items) {
  console.log(item);
}
```

```python
# Python — identical concept, different syntax
for item in items:
    print(item)
```

If you need both index and value, use `enumerate()`:
```python
for i, item in enumerate(items):
    print(i, item)
```

````python
# For loop examples
scores = [0.91, 0.63, 0.78, 0.55, 0.88]

# Basic iteration
for s in scores:
    print(s)

print("---")

# With enumerate for index + value
for i, s in enumerate(scores):
    print(f"Response {i}: {s:.2f}")
````

## 4. `range()`

`range()` generates a sequence of integers. It's the Python equivalent of `for (let i = 0; i < n; i++)`.

| Call | Generates |
|------|-----------|
| `range(5)` | `0, 1, 2, 3, 4` |
| `range(1, 6)` | `1, 2, 3, 4, 5` |
| `range(0, 10, 2)` | `0, 2, 4, 6, 8` |
| `range(10, 0, -1)` | `10, 9, 8, ..., 1` |

The stop value is **exclusive** (same as JavaScript's `Array.from({length: n}, (_, i) => i)`).

````python
# range() examples
print("range(5):")
for i in range(5):
    print(i, end=" ")  # end=" " prints space instead of newline
print()

print("range(1, 6):")
for i in range(1, 6):
    print(i, end=" ")
print()

print("range(0, 10, 2) — every other number:")
for i in range(0, 10, 2):
    print(i, end=" ")
print()

# Convert range to list to see all values
print(list(range(5)))        # [0, 1, 2, 3, 4]
print(list(range(1, 11, 3))) # [1, 4, 7, 10]
````

## 5. `while` Loops

A `while` loop keeps running as long as its condition is `True`. Use it when you don't know in advance how many iterations you need.

```python
while condition:
    # do something
    # make sure condition eventually becomes False!
```

Always ensure the condition will eventually become `False`, or use `break` to exit — otherwise you get an infinite loop.

````python
# while loop example: keep sampling until we have 3 passing scores
import random
random.seed(42)  # for reproducibility

passing_count = 0
attempts = 0

while passing_count < 3:
    score = random.random()  # random float between 0 and 1
    attempts += 1
    if score >= 0.7:
        passing_count += 1
        print(f"  Attempt {attempts}: {score:.3f} — PASS (passing: {passing_count})")
    else:
        print(f"  Attempt {attempts}: {score:.3f} — fail")

print(f"Got 3 passing scores after {attempts} attempts.")
````

## 6. `break` and `continue`

- **`break`**: exit the loop immediately
- **`continue`**: skip the rest of the current iteration and go to the next one

These work the same as in JavaScript.

````python
scores = [0.91, 0.63, 0.78, 0.45, 0.88, 0.33, 0.95]

# continue: skip scores below threshold
print("Scores at or above 0.7:")
for s in scores:
    if s < 0.7:
        continue  # skip this one
    print(f"  {s:.2f}")

print()

# break: stop at the first failing score
print("Scores until first failure (< 0.7):")
for s in scores:
    if s < 0.7:
        print(f"  Found failure: {s:.2f} — stopping")
        break
    print(f"  {s:.2f} — ok")
````

## Why This Matters for AI Research Engineering

Control flow is how you process batches of model outputs:

- **Iterating over responses**: `for response in model_outputs:` — you'll write this hundreds of times
- **Checking conditions**: filtering responses that meet a quality threshold, flagging potential safety issues
- **Counting and accumulating**: tallying correct answers, summing scores across a batch
- **Early stopping**: `break` when you find a critical failure, `continue` to skip malformed inputs

A realistic snippet from an evaluation pipeline:
```python
correct = 0
for response in model_outputs:
    if response["answer"] == response["expected"]:
        correct += 1
accuracy = correct / len(model_outputs)
```

Every evaluation script you write will use loops and conditionals at its core.

## Summary

- **`if`/`elif`/`else`**: Blocks are defined by indentation, not braces. Conditions end with `:`.
- **Comparison operators**: `==`, `!=`, `<`, `>`, `<=`, `>=` — same as JS.
- **Boolean operators**: `and`, `or`, `not` — words, not symbols.
- **`for item in collection`**: Direct iteration, no index needed. Use `enumerate()` if you need both.
- **`range(start, stop, step)`**: Generates integers. `stop` is exclusive.
- **`while condition:`**: Repeats until condition is `False`.
- **`break`**: Exit the loop. **`continue`**: Skip to the next iteration.
