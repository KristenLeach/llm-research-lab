## 1. Variables

A variable is a named container for a value. In Python, you create a variable just by assigning to it — no `let`, `const`, or `var` keyword needed (unlike JavaScript).

```python
x = 10        # JavaScript: let x = 10;
name = "Ada"  # JavaScript: const name = "Ada";
```

Python variables are **dynamically typed**: a variable can hold any type, and you can reassign it to a different type. However, in practice, you should keep variables consistent in type for clarity.

````python
# Assigning variables
model_name = "GPT-4"
score = 0.87
sample_count = 1000
is_passing = True
baseline = None

print(model_name)
print(score)
print(sample_count)
print(is_passing)
print(baseline)
````

## 2. Basic Types

Python has five basic types you'll use constantly in AI research work:

| Type | Example | JS equivalent |
|------|---------|---------------|
| `int` | `42`, `-7`, `0` | `number` (integer) |
| `float` | `0.87`, `3.14`, `-1.5` | `number` (decimal) |
| `str` | `"hello"`, `'world'` | `string` |
| `bool` | `True`, `False` | `boolean` |
| `None` | `None` | `null` |

Notice: Python booleans are capitalized — `True` and `False`, not `true` and `false`.

Use `type()` to inspect what type a value is:

````python
# Checking types
print(type(42))          # <class 'int'>
print(type(0.87))        # <class 'float'>
print(type("GPT-4"))     # <class 'str'>
print(type(True))        # <class 'bool'>
print(type(None))        # <class 'NoneType'>

# You can also use isinstance() to check membership
print(isinstance(42, int))      # True
print(isinstance(0.87, float))  # True
````

## 3. Type Conversion

You frequently need to convert between types — for example, when reading data from a file (everything comes in as a string) or when formatting output.

| Conversion | Function | Example |
|------------|----------|---------|
| String to int | `int()` | `int("42")` → `42` |
| String to float | `float()` | `float("1.5")` → `1.5` |
| Number to string | `str()` | `str(3.14)` → `"3.14"` |
| Anything to bool | `bool()` | `bool(0)` → `False` |

Note: `int()` on a float **truncates** (does not round): `int(3.9)` → `3`.

````python
# Type conversion examples
score_str = "0.87"           # came in as a string
score_float = float(score_str)
print(score_float, type(score_float))  # 0.87 <class 'float'>

count_str = "1000"
count_int = int(count_str)
print(count_int, type(count_int))      # 1000 <class 'int'>

# int() on a float truncates toward zero
print(int(3.9))   # 3
print(int(-3.9))  # -3

# Converting to string (useful for building output)
print(str(42))    # '42'
print(str(True))  # 'True'
````

## 4. F-strings

F-strings (formatted string literals) are the modern Python way to embed values inside strings. Prefix the string with `f` and put expressions in curly braces.

```python
# JavaScript template literal equivalent:
# `The score is ${score.toFixed(2)}`

# Python f-string:
f"The score is {score:.2f}"
```

Common format specifiers:

| Specifier | Meaning | Example |
|-----------|---------|--------|
| `:.2f` | Float with 2 decimal places | `f"{0.8765:.2f}"` → `"0.88"` |
| `:d` | Integer | `f"{42:d}"` → `"42"` |
| `:>10` | Right-align in 10 chars | padding |
| `:.1%` | Percentage with 1 decimal | `f"{0.87:.1%}"` → `"87.0%"` |

````python
# F-string examples
model_name = "GPT-4"
score = 0.8765
sample_count = 1000

# Basic embedding
print(f"Model: {model_name}")

# Float formatting
print(f"Score: {score:.2f}")       # Score: 0.88
print(f"Score: {score:.4f}")       # Score: 0.8765
print(f"Score: {score:.1%}")       # Score: 87.7%

# Integer formatting
print(f"Samples: {sample_count:,}")  # Samples: 1,000  (thousands separator)

# Expressions inside f-strings
correct = 873
total = 1000
print(f"Accuracy: {correct / total:.3f}")
````

## 5. String Methods

Strings in Python have many built-in methods. These are especially useful when cleaning text data — a constant task in AI research.

| Method | What it does | Example |
|--------|-------------|--------|
| `.upper()` | All uppercase | `"hello".upper()` → `"HELLO"` |
| `.lower()` | All lowercase | `"HELLO".lower()` → `"hello"` |
| `.strip()` | Remove leading/trailing whitespace | `"  hi  ".strip()` → `"hi"` |
| `.replace(a, b)` | Replace `a` with `b` | `"cat".replace("c", "b")` → `"bat"` |
| `.split(sep)` | Split into a list | `"a,b,c".split(",")` → `["a","b","c"]` |

Note: strings are **immutable** in Python (same as in JS). These methods return new strings; they don't modify the original.

````python
# String method examples
raw = "  Hello, World!  "

print(raw.strip())            # 'Hello, World!'
print(raw.strip().lower())    # 'hello, world!'  (chaining methods)
print(raw.strip().upper())    # 'HELLO, WORLD!'

# replace
label = "model_a_results"
print(label.replace("_", "-"))   # 'model-a-results'

# split
csv_row = "gpt-4,0.87,1000"
parts = csv_row.split(",")
print(parts)        # ['gpt-4', '0.87', '1000']
print(parts[0])     # 'gpt-4'
print(parts[1])     # '0.87'
````

## Why This Matters for AI Research Engineering

In AI research, you constantly work with:

Knowing how to store, convert, and format these values is the foundation of every analysis script you'll write. When you pull data from a JSON API response, everything comes back as strings or numbers. You'll type-convert, format with f-strings, and clean strings with `.strip()` and `.split()` constantly.

A typical line from an evaluation pipeline:
```python
print(f"[{model_name}] accuracy={accuracy:.3f} on {n_samples:,} samples")
```

## Summary

- **Variables** are created by assignment: `x = 10`. No keyword needed.
- **Five basic types**: `int` (whole numbers), `float` (decimals), `str` (text), `bool` (`True`/`False`), `None` (absence of value).
- **`type(x)`** returns the type of `x`; **`isinstance(x, T)`** checks if `x` is of type `T`.
- **Type conversion**: `int()`, `float()`, `str()`, `bool()`. Conversions can raise `ValueError` if the string isn't valid (e.g., `int("hello")`).
- **F-strings**: `f"text {variable:.2f}"` — the cleanest way to embed values in strings.
- **String methods** return new strings (strings are immutable). Chain them: `raw.strip().lower().split()`.
