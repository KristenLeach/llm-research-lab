## Exercise 1: Assign variables and use an f-string

Assign a model name and score, then format them into a readable string using an f-string.
The `:.2f` format specifier rounds the float to 2 decimal places — handy for displaying scores cleanly.

```python
# Assign the model name (a string) and score (a float)
model_name = "GPT-4"
score = 0.87

# f-strings let you embed expressions inside curly braces.
# :.2f means: format as a float with 2 decimal places.
result = f"{model_name} scored {score:.2f}"

print(result)  # GPT-4 scored 0.87
```

## Exercise 2: Type conversion

Convert a string to an integer, then multiply it. This comes up constantly when reading data
from files or APIs — values often arrive as strings even when they represent numbers.

```python
# int() converts a string to an integer.
# Without the conversion, "42" * 2 would give "4242" (string repetition!).
result = int("42") * 2

print(result)        # 84
print(type(result))  # <class 'int'>
```

## Exercise 3: String methods — strip and split

Chain `.strip()` and `.split()` to clean and tokenize a string.
- `.strip()` removes leading and trailing whitespace (like JS's `.trim()`)
- `.split()` with no arguments splits on any whitespace and discards empty strings

```python
messy = "  hello world  "

# .strip() removes whitespace from both ends — like JS's .trim()
# .split() with no argument splits on any whitespace run
words = messy.strip().split()

print(words)         # ['hello', 'world']
print(len(words))    # 2

# Why chain them? strip() first ensures no empty strings from leading/trailing spaces.
# Try messy.split() without strip() — it still works here,
# but with split(" ") you'd get empty strings at the edges.
```
