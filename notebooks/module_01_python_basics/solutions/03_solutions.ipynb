## Exercise 1: `classify_score(score)`

A function that takes a float score and returns a string label.
This is a pure function — same input always gives same output, no side effects.
That's the ideal for evaluation helpers.

```python
def classify_score(score):
    """Return a category label for a numeric score between 0 and 1."""
    # Use elif (not else if) for chained conditions.
    if score >= 0.9:
        return "excellent"
    elif score >= 0.7:
        return "good"
    elif score >= 0.5:
        return "fair"
    else:
        return "poor"


# Test it with a few values
print(classify_score(0.95))  # excellent
print(classify_score(0.85))  # good
print(classify_score(0.60))  # fair
print(classify_score(0.30))  # poor
```

## Exercise 2: `format_result(model_name, score, task="general")`

A function with a **default argument**. In Python, default arguments go at the end of the
parameter list (just like in JS). The caller can omit `task` and get `"general"` automatically.

```python
def format_result(model_name, score, task="general"):
    """Return a formatted string summarising a model's score on a task."""
    # classify_score is defined above — functions can call other functions.
    label = classify_score(score)
    # :.2f formats the float to 2 decimal places.
    return f"{model_name} | {task} | {score:.2f} ({label})"


# Call without the optional argument — uses default task="general"
print(format_result("GPT-4", 0.87))
# GPT-4 | general | 0.87 (good)

# Call with all three arguments
print(format_result("Claude", 0.94, task="harmful_refusal"))
# Claude | harmful_refusal | 0.94 (excellent)
```

## Exercise 3: `count_above_threshold(scores, threshold=0.5)`

A function that accepts a list and an optional threshold, returns an integer count.
The default `threshold=0.5` makes this convenient to call without specifying a cutoff every time,
but still flexible when you need a different value.

```python
def count_above_threshold(scores, threshold=0.5):
    """Count how many values in scores are strictly above threshold."""
    # sum() with a generator expression is a concise Pythonic pattern.
    # For each score, (s > threshold) evaluates to True (1) or False (0).
    # sum() adds them up — giving the count of True values.
    return sum(1 for s in scores if s > threshold)


sample_scores = [0.92, 0.45, 0.78, 0.88, 0.61, 0.95, 0.33]

# Using default threshold (0.5)
print(count_above_threshold(sample_scores))         # 5

# Using a custom threshold
print(count_above_threshold(sample_scores, 0.8))    # 3
print(count_above_threshold(sample_scores, 0.9))    # 2

# You can also pass threshold as a keyword argument (more readable)
print(count_above_threshold(sample_scores, threshold=0.7))  # 4
```
