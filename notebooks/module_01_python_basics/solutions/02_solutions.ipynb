## Exercise 1: if/elif/else — categorize a score

Use an if/elif/else chain to map a numeric score to a human-readable category.
Python uses `elif` (not `else if` like JS). Conditions are checked top to bottom — the first
one that's `True` wins, and the rest are skipped.

```python
score = 0.85

# Python uses elif, not else if.
# Conditions are evaluated top-to-bottom; only the first True branch runs.
if score >= 0.9:
    category = "excellent"
elif score >= 0.7:
    category = "good"
elif score >= 0.5:
    category = "fair"
else:
    category = "poor"

print(f"Score {score} is: {category}")  # Score 0.85 is: good

# Try changing score to 0.95, 0.6, 0.3 to see each branch fire.
```

## Exercise 2: for loop — count items above threshold

Loop through a list and count how many values exceed a threshold.
This pattern is fundamental in evaluation: "how many model outputs scored above 0.8?"

```python
scores = [0.92, 0.45, 0.78, 0.88, 0.61, 0.95, 0.33]
threshold = 0.8

# Initialize a counter before the loop — a common Python pattern.
count = 0
for s in scores:
    if s > threshold:
        count += 1  # Same as count = count + 1

print(f"Scores above {threshold}: {count}")  # Scores above 0.8: 3

# Pythonic alternative using sum() + a generator expression:
count_alt = sum(1 for s in scores if s > threshold)
print(f"(Alternative) Scores above {threshold}: {count_alt}")
```

## Exercise 3: for loop with range — print every other number

`range(start, stop, step)` generates a sequence of integers.
The `step` argument controls the increment — use `2` to skip every other number.
Note: `stop` is **exclusive** (like JS array slicing).

```python
# range(start, stop, step)
# - start: 0 (inclusive)
# - stop: 11 (exclusive — so we go up to 10)
# - step: 2 (skip every other number)
for i in range(0, 11, 2):
    print(i)

# Output: 0, 2, 4, 6, 8, 10

# You can also collect results into a list:
every_other = list(range(0, 11, 2))
print(every_other)  # [0, 2, 4, 6, 8, 10]
```
