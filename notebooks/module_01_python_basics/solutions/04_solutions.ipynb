## Exercise 1: List comprehension — filter scores above 0.8

List comprehensions are one of Python's most useful features.
They replace a for-loop-with-append in a single, readable line.
Think of them as: `[expression for item in iterable if condition]`.

```python
scores = [0.92, 0.45, 0.78, 0.88, 0.61, 0.95, 0.33, 0.81]

# List comprehension: keep only scores strictly above 0.8
# Read as: "give me s, for each s in scores, but only if s > 0.8"
high_scores = [s for s in scores if s > 0.8]

print(high_scores)  # [0.92, 0.88, 0.95, 0.81]

# Equivalent for-loop version (longer, same result):
# high_scores = []
# for s in scores:
#     if s > 0.8:
#         high_scores.append(s)
```

## Exercise 2: Dict from two parallel lists

`zip()` pairs up two iterables element-by-element — like a zipper on a jacket.
`dict()` then turns those pairs into key-value entries.
This is useful when you have separate lists of keys and values (e.g., from a CSV header and a row).

```python
models = ["model-a", "model-b", "model-c"]
scores = [0.92, 0.74, 0.85]

# zip() pairs elements by position: ("model-a", 0.92), ("model-b", 0.74), ...
# dict() turns those pairs into a dictionary.
model_scores = dict(zip(models, scores))

print(model_scores)
# {'model-a': 0.92, 'model-b': 0.74, 'model-c': 0.85}

# Access a specific model's score:
print(model_scores["model-a"])  # 0.92

# Dict comprehension alternative (also valid):
model_scores_alt = {m: s for m, s in zip(models, scores)}
print(model_scores_alt == model_scores)  # True
```

## Exercise 3: Set for unique values, convert to sorted list

A `set` automatically deduplicates — any duplicate values are silently dropped.
Sets are unordered, so convert to a sorted list when you need a predictable order.

```python
categories = [
    "harmful_content",
    "misinformation",
    "harmful_content",
    "jailbreak",
    "misinformation",
    "misinformation",
]

# set() removes duplicates — order is not guaranteed
unique_set = set(categories)
print(unique_set)
# e.g. {'harmful_content', 'jailbreak', 'misinformation'} (order may vary)

# sorted() returns a new sorted list (alphabetical for strings)
unique_sorted = sorted(unique_set)
print(unique_sorted)
# ['harmful_content', 'jailbreak', 'misinformation']

# One-liner combining both steps:
unique_sorted2 = sorted(set(categories))
print(unique_sorted2 == unique_sorted)  # True
```

## Exercise 4: Dict comprehension — pass/fail

A **dict comprehension** builds a dictionary in a single expression, just like a list
comprehension builds a list. The syntax is `{key: value for item in iterable}`.
Here we map each model name to a boolean: did it pass the threshold?

```python
model_scores = {
    "model-a": 0.92,
    "model-b": 0.74,
    "model-c": 0.85,
    "model-d": 0.61,
}
pass_threshold = 0.8

# Dict comprehension: for each model, record True if score >= threshold, else False.
# .items() gives (key, value) pairs from the dict — like Object.entries() in JS.
pass_fail = {model: score >= pass_threshold for model, score in model_scores.items()}

print(pass_fail)
# {'model-a': True, 'model-b': False, 'model-c': True, 'model-d': False}

# Filter to just the passing models:
passing = [m for m, passed in pass_fail.items() if passed]
print(f"Passing models: {passing}")  # Passing models: ['model-a', 'model-c']
```
