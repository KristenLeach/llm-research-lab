## 1. Defining Functions

Functions let you name and reuse a block of code. The syntax uses `def`, followed by the function name, parentheses with parameters, and a colon.

```javascript
// JavaScript
function greet(name) {
  return `Hello, ${name}!`;
}
```

```python
# Python
def greet(name):
    return f"Hello, {name}!"
```

Key points:
- Use `def` keyword (not `function`)
- No curly braces — body is indented
- `return` works the same as JS
- A function without `return` implicitly returns `None`

````python
# Basic function definition and calling
def greet(name):
    return f"Hello, {name}!"

print(greet("Ada"))     # Hello, Ada!
print(greet("Claude"))  # Hello, Claude!

# A function that takes two arguments
def add(a, b):
    return a + b

result = add(3, 5)
print(result)  # 8
````

## 2. Default Parameter Values

Parameters can have default values, making them optional when calling the function. Parameters with defaults must come **after** parameters without defaults.

```python
def func(required, optional="default_value"):
    ...
```

You can also call functions with **keyword arguments** — naming the argument explicitly. This makes calls more readable and lets you pass arguments out of order:

```python
func(required="x", optional="y")
func(optional="y", required="x")  # same result
```

````python
# Default parameters
def classify_score(score, threshold=0.7):
    if score >= threshold:
        return "pass"
    else:
        return "fail"

# Using default threshold (0.7)
print(classify_score(0.85))         # pass
print(classify_score(0.65))         # fail

# Overriding the default
print(classify_score(0.65, 0.6))    # pass (threshold is now 0.6)

# Using keyword argument — more readable
print(classify_score(0.65, threshold=0.6))  # pass
````

## 3. Docstrings

A docstring is a string literal that appears as the first statement in a function. It documents what the function does. Python tools and IDEs use docstrings to provide help text.

```python
def my_function(x):
    """One-line description of what this does."""
    ...
```

For longer documentation, use triple quotes with multiple lines:

```python
def my_function(x, y):
    """
    Description of what this function does.

    Args:
        x: description of x
        y: description of y

    Returns:
        description of return value
    """
    ...
```

Access the docstring with `help(function_name)` or `function_name.__doc__`.

````python
def evaluate_response(response_text, expected_answer, threshold=0.7):
    """
    Evaluate whether a model response matches the expected answer.

    Args:
        response_text: The model's response string.
        expected_answer: The correct answer string.
        threshold: Unused here, but included to show pattern. Default 0.7.

    Returns:
        True if the response matches the expected answer (case-insensitive), False otherwise.
    """
    return response_text.strip().lower() == expected_answer.strip().lower()

# Try it
print(evaluate_response("Paris", "paris"))        # True
print(evaluate_response("london", "Paris"))       # False
print(evaluate_response("  Paris  ", "Paris"))    # True (strips whitespace)

# View the docstring
help(evaluate_response)
````

## 4. Multiple Return Values

Python functions can return multiple values using **tuples**. A tuple is just comma-separated values (parentheses are optional but conventional).

```python
def get_stats(values):
    return min(values), max(values)  # returns a tuple

low, high = get_stats([1, 5, 3])  # "unpacking" the tuple
```

This is similar to destructuring assignment in JavaScript:
```javascript
const [low, high] = getStats([1, 5, 3]);
```

````python
# Multiple return values
def score_summary(scores):
    """Return (count, mean, passing_count) for a list of scores."""
    count = len(scores)
    mean = sum(scores) / count
    passing = sum(1 for s in scores if s >= 0.7)
    return count, mean, passing  # returns a tuple

scores = [0.91, 0.63, 0.78, 0.55, 0.88]

# Unpack all three values at once
n, avg, passed = score_summary(scores)
print(f"Count: {n}")
print(f"Mean:  {avg:.3f}")
print(f"Passed: {passed}/{n}")

# Or capture as a single tuple
result = score_summary(scores)
print(type(result))   # <class 'tuple'>
print(result)         # (5, 0.75, 3)
````

## 5. Variable Scope

Variables created **inside** a function are **local** — they only exist during that function call and are invisible outside it.

Variables created at the **module level** (outside any function) are **global** to the module and can be read inside functions.

```python
x = 10  # module-level (global)

def my_func():
    y = 20       # local — only exists here
    print(x)     # OK — can read global x
    print(y)     # OK — local

my_func()
print(x)         # OK
print(y)         # NameError — y doesn't exist here
```

Best practice: **pass data into functions as arguments** rather than relying on global variables. This makes functions predictable and testable.

````python
# Scope example
DEFAULT_THRESHOLD = 0.7  # module-level constant (convention: UPPER_CASE)

def is_passing(score):
    # Can read DEFAULT_THRESHOLD from module level
    result = score >= DEFAULT_THRESHOLD  # 'result' is local
    return result

print(is_passing(0.8))   # True
print(is_passing(0.6))   # False

# 'result' from inside the function is NOT accessible here
# Uncomment to see the error:
print(result)  # NameError: name 'result' is not defined
````

## 6. A Brief Look at `*args`

You'll often see `*args` in Python code. It allows a function to accept **any number of positional arguments**, collecting them into a tuple.

```python
def my_func(*args):
    for arg in args:
        print(arg)

my_func(1, 2, 3)      # prints 1, 2, 3
my_func("a", "b")     # prints a, b
```

You don't need to write `*args` yourself yet, but you'll encounter it in libraries, so it's useful to recognize it. There's also `**kwargs` for variable keyword arguments — we'll cover those in a later module.

````python
# *args example — accepts any number of scores
def average(*scores):
    """Compute the mean of any number of scores."""
    if not scores:
        return 0.0
    return sum(scores) / len(scores)

print(average(0.8, 0.9, 0.7))               # 0.8
print(average(0.6, 0.7, 0.8, 0.9, 1.0))    # 0.8
print(average(0.5))                          # 0.5

# Inside the function, 'scores' is just a tuple
def show_args(*args):
    print(type(args), args)

show_args(1, 2, 3)   # <class 'tuple'> (1, 2, 3)
````

## Summary

- **`def name(params):`** defines a function. Body is indented. `return` sends a value back.
- **Default parameters**: `def f(x, y=10)` — `y` is optional. Defaults come after required params.
- **Keyword arguments**: `f(x=1, y=2)` — name them explicitly for clarity and order-independence.
- **Docstrings**: triple-quoted string as the first line of a function. Documents purpose, args, return value.
- **Multiple returns**: `return a, b` returns a tuple; unpack with `x, y = f()`.
- **Scope**: local variables live inside the function. Module-level variables are readable everywhere. Prefer passing data as arguments.
- **`*args`**: collects any number of positional arguments into a tuple. You'll see it in libraries.
