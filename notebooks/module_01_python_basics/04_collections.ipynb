## 1. Lists

A list is an ordered, mutable sequence — the Python equivalent of a JavaScript array.

```javascript
// JavaScript
const scores = [0.91, 0.63, 0.78];
scores.push(0.85);   // append
scores[0];           // 0.91
scores.length;       // 4
```

```python
# Python
scores = [0.91, 0.63, 0.78]
scores.append(0.85)  # append
scores[0]            # 0.91
len(scores)          # 4
```

Python uses negative indices to count from the end: `scores[-1]` is the last element.

````python
# List creation and basic operations
scores = [0.91, 0.63, 0.78, 0.55, 0.88]

# Indexing
print(scores[0])    # 0.91  (first)
print(scores[-1])   # 0.88  (last)
print(scores[-2])   # 0.55  (second to last)

# Slicing: [start:stop] — stop is exclusive
print(scores[1:3])  # [0.63, 0.78]
print(scores[:2])   # [0.91, 0.63]  (first 2)
print(scores[2:])   # [0.78, 0.55, 0.88]  (from index 2 to end)
print(scores[::2])  # [0.91, 0.78, 0.88]  (every other)

# Length and membership
print(len(scores))           # 5
print(0.63 in scores)        # True
print(0.99 in scores)        # False

# Mutating the list
scores.append(0.72)          # add to end
print(scores)                # [..., 0.72]

scores.extend([0.81, 0.67])  # add multiple items
print(len(scores))           # 8

scores.remove(0.55)          # remove by value (first occurrence)
print(0.55 in scores)        # False
````

## 2. Dictionaries

A dictionary maps keys to values. It's the Python equivalent of a JavaScript object or `Map`.

```javascript
// JavaScript object
const result = { model: "gpt-4", score: 0.87, samples: 1000 };
result.model;            // 'gpt-4'
result["score"];         // 0.87
```

```python
# Python dict
result = {"model": "gpt-4", "score": 0.87, "samples": 1000}
result["model"]          # 'gpt-4'
result.get("score")      # 0.87
```

Use `.get(key, default)` instead of `dict[key]` when the key might not exist — it returns `None` (or your default) instead of raising a `KeyError`.

````python
# Dictionary creation and access
result = {"model": "gpt-4", "score": 0.87, "samples": 1000}

# Access by key
print(result["model"])            # 'gpt-4'
print(result["score"])            # 0.87

# Safe access with .get()
print(result.get("samples"))      # 1000
print(result.get("task"))         # None (key doesn't exist)
print(result.get("task", "N/A")) # 'N/A' (custom default)

# Adding and updating
result["task"] = "general"        # add new key
result["score"] = 0.90            # update existing key
print(result)

# Iterating
print("\nKeys:", list(result.keys()))
print("Values:", list(result.values()))

print("\nKey-value pairs:")
for key, value in result.items():
    print(f"  {key}: {value}")
````

## 3. Sets

A set is an **unordered** collection of **unique** values — like a mathematical set. Sets are great for:
- Removing duplicates from a list
- Fast membership testing
- Finding overlap or difference between two groups

| Operation | Syntax | Meaning |
|-----------|--------|---------|
| Union | `a \| b` | all items in a or b |
| Intersection | `a & b` | items in both a and b |
| Difference | `a - b` | items in a but not b |

````python
# Sets — uniqueness and set operations
models_a = {"gpt-4", "claude-3", "gemini-pro"}
models_b = {"claude-3", "gemini-pro", "llama-3"}

print("Union:", models_a | models_b)          # all models
print("Intersection:", models_a & models_b)   # shared models
print("In A but not B:", models_a - models_b) # only in A

# Add an item
models_a.add("mixtral")
print("After add:", models_a)

# Membership test (very fast, even for large sets)
print("gpt-4" in models_a)    # True
print("llama-3" in models_a)  # False

# Deduplication: convert list → set → back to list
tags = ["safety", "factuality", "safety", "math", "factuality", "safety"]
unique_tags = list(set(tags))
print("Unique tags:", sorted(unique_tags))  # sorted for consistent order
````

## 4. Tuples

A tuple is an **ordered, immutable** sequence. Once created, you can't add, remove, or change elements. Tuples are useful for:
- Returning multiple values from a function (as we saw in Notebook 3)
- Representing fixed records (like a coordinate or a key-value pair)
- Dictionary keys (since tuples are hashable, unlike lists)

Create tuples with parentheses (or just commas):
```python
point = (3.0, 4.0)
point = 3.0, 4.0   # same thing
```

````python
# Tuples
coords = (48.8566, 2.3522)   # latitude, longitude of Paris

print(coords[0])    # 48.8566
print(coords[1])    # 2.3522
print(len(coords))  # 2

# Unpacking (very common pattern)
lat, lon = coords
print(f"Latitude: {lat}, Longitude: {lon}")

# Tuples from functions
def min_max(values):
    return min(values), max(values)

scores = [0.91, 0.63, 0.78, 0.55, 0.88]
lo, hi = min_max(scores)
print(f"Range: {lo:.2f} – {hi:.2f}")

# Immutability — this would raise a TypeError:
# coords[0] = 0.0  # TypeError: 'tuple' object does not support item assignment
````

## 5. List Comprehensions

A list comprehension builds a new list by applying an expression to each item in an iterable. It's the Pythonic alternative to a `for` loop that builds a list.

```python
# These two are equivalent:

# Loop version
result = []
for x in numbers:
    result.append(x * 2)

# List comprehension
result = [x * 2 for x in numbers]
```

You can add a filter condition:
```python
# Only include positive numbers, doubled
result = [x * 2 for x in numbers if x > 0]
```

This is similar to JavaScript's `.map()` and `.filter()` chained together.

````python
# List comprehension examples
scores = [0.91, 0.63, 0.78, 0.55, 0.88, 0.45, 0.82]

# Transform: multiply each score by 100 (convert to percentage)
percentages = [s * 100 for s in scores]
print("Percentages:", percentages)

# Filter: only keep scores above 0.7
passing = [s for s in scores if s > 0.7]
print("Passing (>0.7):", passing)

# Transform + filter: percentages for passing scores only
passing_pct = [s * 100 for s in scores if s > 0.7]
print("Passing percentages:", passing_pct)

# Classify each score
labels = ["pass" if s >= 0.7 else "fail" for s in scores]
print("Labels:", labels)
````

## 6. Dict Comprehensions

Dict comprehensions build a new dictionary, similar to list comprehensions but with `key: value` syntax:

```python
{key_expr: value_expr for item in iterable}
{key_expr: value_expr for item in iterable if condition}
```

A common pattern: building a dict from two parallel lists using `zip()`:
```python
keys = ["a", "b", "c"]
values = [1, 2, 3]
d = {k: v for k, v in zip(keys, values)}  # {"a": 1, "b": 2, "c": 3}
```

````python
# Dict comprehension examples
models = ["gpt-4", "claude-3", "gemini-pro"]
scores = [0.91, 0.88, 0.79]

# Build model → score dict
model_scores = {model: score for model, score in zip(models, scores)}
print("Model scores:", model_scores)

# Build model → label dict (classify each)
model_labels = {m: ("pass" if s >= 0.8 else "fail") for m, s in zip(models, scores)}
print("Labels:", model_labels)

# Filter: only include passing models
passing_models = {m: s for m, s in zip(models, scores) if s >= 0.8}
print("Passing models:", passing_models)

# Transform an existing dict
raw_scores = {"gpt-4": 0.91, "claude-3": 0.88}
rounded = {m: round(s, 1) for m, s in raw_scores.items()}
print("Rounded:", rounded)
````

## Why This Matters for AI Research Engineering

Lists and dicts are the data structures of AI research:

- **Model outputs** come back as lists of dicts: `[{"id": 1, "response": "...", "score": 0.87}, ...]`
- **Evaluation results** are stored in dicts: `{"gpt-4": 0.91, "claude-3": 0.88}`
- **Filtering bad outputs**: `[r for r in responses if not r["flagged"]]`
- **Grouping by model**: build a dict of `{model: [responses]}` in one comprehension
- **Deduplicating**: sets remove duplicate prompts, test cases, or labels instantly

You'll write list comprehensions daily when processing batches of model outputs. They're one of the most distinctively Python patterns — once you internalize them, your code becomes both shorter and more readable than the equivalent JS `.map().filter()` chains.

## Summary

- **Lists** `[a, b, c]`: ordered, mutable. Index with `[i]`, slice with `[start:stop]`, append with `.append()`, extend with `.extend()`, test membership with `in`.
- **Dicts** `{key: value}`: key-value store. Access with `d[key]` or safe `d.get(key, default)`. Iterate with `.items()`. Add/update with `d[key] = value`.
- **Sets** `{a, b, c}`: unordered, unique values. Union `|`, intersection `&`, difference `-`. Great for deduplication and membership testing.
- **Tuples** `(a, b)`: ordered, **immutable**. Unpack with `a, b = tuple`. Returned from functions.
- **List comprehensions** `[expr for x in iterable if cond]`: concise list building. Replaces `.map().filter()`.
- **Dict comprehensions** `{k: v for k, v in iterable}`: concise dict building. Use `zip()` to pair parallel lists.
