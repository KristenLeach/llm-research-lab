# Module 06 — Writing Reusable Scripts

Notebooks are great for exploration, but production research code lives in `.py` scripts.
This module teaches you how to write Python scripts that can be run from the command line,
imported by other scripts, and shared with colleagues.

---

## Learning Objectives

By the end of this module, you will be able to:

- Structure a Python script with functions and a `main()` entry point
- Use `if __name__ == "__main__":` to make scripts both importable and runnable
- Add command-line argument parsing with `argparse`
- Use the `logging` module instead of print statements for production code
- Import your own modules from other files
- Understand the difference between a module, a package, and a script

---

## What You'll Build

**Mini-project:** Write a script `run_evaluation.py` that:
- Accepts `--input` (path to JSON), `--output` (path to results JSON), and `--threshold` (float) as CLI args
- Loads model outputs, filters by threshold, computes summary statistics
- Saves results to the output path
- Logs progress to stdout using `logging`
- Can be imported and its functions called from other scripts

## Key Concepts

- **`__name__`**: a special variable — equals `"__main__"` when the file is run directly, or the module name when imported
- **`argparse`**: the standard library module for parsing command-line arguments
- **`logging`**: structured logging with severity levels (DEBUG, INFO, WARNING, ERROR)
- **Module**: any `.py` file; import it with `import mymodule` or `from mymodule import func`
- **Package**: a directory containing an `__init__.py` file and one or more modules
- **`sys.argv`**: the raw list of command-line arguments (argparse wraps this)
- **Entry point**: the `main()` function — the starting point of execution
