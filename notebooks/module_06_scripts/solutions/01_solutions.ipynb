## Exercise 1 Solution

```python
sample_outputs = [
    {"id": 1, "model": "model-a", "flagged": False},
    {"id": 2, "model": "model-a", "flagged": False},
    {"id": 3, "model": "model-b", "flagged": True},
    {"id": 4, "model": "model-b", "flagged": True},
    {"id": 5, "model": "model-b", "flagged": False},
]

def get_model_flag_rates(outputs):
    # Group outputs by model, then compute flag rate per group
    from collections import defaultdict
    counts = defaultdict(lambda: {"total": 0, "flagged": 0})
    for o in outputs:
        counts[o["model"]]["total"] += 1
        counts[o["model"]]["flagged"] += int(o["flagged"])
    return {m: c["flagged"] / c["total"] for m, c in counts.items()}

rates  = get_model_flag_rates(sample_outputs)
b_rate = round(float(rates["model-b"]), 2)
```

```python
check_type(rates, dict, "rates is a dict")
check_contains(rates, "model-a", "rates has model-a")
check_contains(rates, "model-b", "rates has model-b")
check_approx(rates["model-a"], 0.0, 1e-6, "model-a flag rate is 0.0")
check_approx(b_rate, 0.67, 0.01, "model-b flag rate is 0.67")
```

## Exercise 2 Solution

```python
%%writefile scripts/model_rates.py
import json
from collections import defaultdict
from pathlib import Path

DATA_PATH = Path("../../data/synthetic/model_outputs.json")


def get_model_flag_rates(outputs):
    counts = defaultdict(lambda: {"total": 0, "flagged": 0})
    for o in outputs:
        counts[o["model"]]["total"] += 1
        counts[o["model"]]["flagged"] += int(o["flagged"])
    return {m: c["flagged"] / c["total"] for m, c in counts.items()}


def main():
    with open(DATA_PATH, encoding="utf-8") as f:
        outputs = json.load(f)
    rates = get_model_flag_rates(outputs)
    for model, rate in sorted(rates.items()):
        print(f"{model}: {rate:.1%}")


if __name__ == "__main__":
    main()
```
