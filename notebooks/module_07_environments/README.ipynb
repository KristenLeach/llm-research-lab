# Module 07 — Environments, Dependencies & Reproducibility

In AI research, reproducibility is a first-class concern. Another researcher (or future you)
should be able to run your code from scratch and get the same results.
This module covers Python's environment and dependency management tools.

If you're coming from Node.js: virtual environments are like `node_modules` isolation,
`uv` or `pip` is like `npm`, and `pyproject.toml` or `requirements.txt` is like `package.json`.

---

## Learning Objectives

By the end of this module, you will be able to:

- Understand what a virtual environment is and why it matters
- Create and activate a virtual environment with `uv venv` or `python -m venv`
- Install and manage packages with `uv pip install` or `pip install`
- Pin exact dependency versions with `requirements.txt` or `pyproject.toml`
- Set random seeds to make stochastic code reproducible
- Write a reproducibility checklist / README so others can reproduce your results

---

## What You'll Build

**Mini-project:** Create a reproducible analysis setup:
- A virtual environment with pinned dependencies
- A `requirements.txt` (or `pyproject.toml`) for the project
- A script that sets all random seeds at the top
- A `REPRODUCE.md` with step-by-step instructions
- Verify that running the analysis twice gives identical results

## Key Concepts

- **Virtual environment**: an isolated Python installation with its own packages
- **`uv`**: a fast modern Python package manager (recommended)
- **`pip`**: the classic package installer (still widely used)
- **`requirements.txt`**: a file listing exact package versions to install
- **`pyproject.toml`**: the modern Python project configuration file
- **Random seed**: `random.seed(42)`, `numpy.random.seed(42)` — makes randomness reproducible
- **Determinism**: given the same inputs and seeds, code should produce the same outputs
