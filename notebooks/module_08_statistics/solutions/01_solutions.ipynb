## Exercise 1 Solution — Compute Mean and Median

```python
import statistics

model_scores = [0.73, 0.81, 0.69, 0.77, 0.85, 0.72, 0.80, 0.68, 0.79, 0.76]

scores_mean = round(float(statistics.mean(model_scores)), 4)
scores_median = round(float(statistics.median(model_scores)), 4)

print(f"Mean:   {scores_mean}")
print(f"Median: {scores_median}")
```

## Exercise 3 Solution — Detect Skew

```python
import statistics

skewed_scores = [0.88, 0.91, 0.87, 0.90, 0.89, 0.92, 0.86, 0.90, 0.23, 0.18]

skew_mean = round(float(statistics.mean(skewed_scores)), 4)
skew_median = round(float(statistics.median(skewed_scores)), 4)

# mean < median means left-skewed (low outliers pulled the mean down)
if skew_mean < skew_median - 0.01:
    skew_direction = "left"
elif skew_mean > skew_median + 0.01:
    skew_direction = "right"
else:
    skew_direction = "symmetric"

print(f"Mean: {skew_mean}  Median: {skew_median}  Skew: {skew_direction}")
```

```python
check_approx(skew_mean, 0.754, 0.001, "skew_mean is correct")
check_approx(skew_median, 0.885, 0.001, "skew_median is correct")
check_equal(skew_direction, "left", "skew_direction is 'left' (mean < median)")
```

## Exercise 4 Solution — Build a Quick Summary Function

```python
import statistics

def score_summary(scores):
    """Return a dict with mean, median, and skew direction for a list of scores."""
    mean   = round(float(statistics.mean(scores)), 4)
    median = round(float(statistics.median(scores)), 4)

    if abs(mean - median) < 0.01:
        skew = "symmetric"
    elif mean < median:
        skew = "left"
    else:
        skew = "right"

    return {"mean": mean, "median": median, "skew": skew}


sample = [0.88, 0.91, 0.87, 0.90, 0.89, 0.92, 0.86, 0.90, 0.23, 0.18]
result = score_summary(sample)
print(result)
```
