# Module 08 — Basic Statistics for Experiments

AI research requires you to distinguish real differences from random noise.
This module covers the statistical tools you'll use most often: computing descriptive statistics,
understanding uncertainty, and deciding whether an observed difference is meaningful.

You don't need a statistics background — just enough to reason carefully about your results
and not overstate your conclusions.

---

## Learning Objectives

By the end of this module, you will be able to:

- Compute mean, median, standard deviation, and quantiles in Python
- Understand what a confidence interval means and how to compute one
- Interpret a p-value correctly (and understand its limitations)
- Understand effect size: is the difference large enough to matter?
- Avoid common statistical mistakes in AI evaluation (p-hacking, underpowered tests)
- Use `scipy.stats` for basic hypothesis tests

## Key Concepts

- **Mean vs. median**: mean is sensitive to outliers; median is more robust
- **Standard deviation**: how spread out the values are
- **Standard error**: how uncertain your estimate of the mean is
- **Confidence interval**: a range that likely contains the true value
- **p-value**: probability of observing this result (or more extreme) if there were no real effect
- **Effect size**: the magnitude of the difference, not just statistical significance
- **Bootstrap**: a resampling technique for estimating uncertainty
- **Type I error**: false positive (concluding there's a difference when there isn't)
- **Type II error**: false negative (missing a real difference)
