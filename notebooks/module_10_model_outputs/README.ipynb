# Module 10 — Analysis of Model Outputs

In practice, AI research engineering means loading thousands of model responses,
computing statistics, identifying patterns, and surfacing interesting findings.
This module focuses on the hands-on analysis workflow for a batch of model outputs.

---

## Learning Objectives

By the end of this module, you will be able to:

- Load and inspect a large batch of model responses from JSON
- Compute summary statistics over model outputs (response length, score distribution, flag rate)
- Identify patterns: what kinds of prompts get flagged? What models score highest?
- Build simple heuristic classifiers (keyword matching, length-based, score-based)
- Compare outputs across multiple models on the same prompts
- Slice results by metadata (task type, model version, date)

---

## What You'll Build

**Mini-project:** Deep analysis of `model_outputs.json`:
- Overall statistics: how many outputs, what fraction are flagged, score distribution
- Per-model breakdown: flag rate, average score, output length by model
- Pattern finding: what makes an output likely to be flagged? (heuristics)
- Comparison: for the same prompts, how do different models differ?
- Summary findings: 3-5 bullet points describing what you found

## Key Concepts

- **Batch analysis**: processing many outputs at once rather than one at a time
- **Flag rate**: the fraction of outputs marked as concerning
- **Heuristic**: a rule-of-thumb classifier (fast but imperfect)
- **Slice analysis**: computing metrics for subgroups (by model, task type, etc.)
- **Outlier detection**: finding unusually high or low scoring outputs for qualitative review
- **Comparison table**: side-by-side metrics for multiple models
