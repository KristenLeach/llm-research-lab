# Module 10 — Notebook 1 Solutions: Loading and Inspecting

```
import json
import sys
from pathlib import Path

sys.path.insert(0, "../../../")
from src.checks import check_equal, check_type, check_contains, check_approx, check_length
```

## Exercise 1 — Solution: Load the JSON File

```
data_path = Path("../../../data/synthetic/model_outputs.json")
with open(data_path) as f:
    outputs = json.load(f)

print(f"Loaded {len(outputs)} records")
print(json.dumps(outputs[0], indent=2))
```

```
# Check Exercise 1
check_equal(len(outputs), 20, "outputs has 20 records")
check_type(outputs, list, "outputs is a list")
check_type(outputs[0], dict, "first element is a dict")
```

## Exercise 2 — Solution: Count Flagged Outputs and Compute Flag Rate

```
flagged_count = sum(1 for r in outputs if r['flagged'])
flag_rate = round(flagged_count / len(outputs), 4)

print(f"Flagged: {flagged_count}")
print(f"Flag rate: {flag_rate}")
```

```
# Check Exercise 2
check_equal(flagged_count, 7, "flagged_count is 7")
check_approx(flag_rate, 0.35, 0.001, "flag_rate is approximately 0.35")
```

## Exercise 3 — Solution: List Unique Models

```
models = sorted(set(r['model'] for r in outputs))
print(f"Models: {models}")
```

```
# Check Exercise 3
check_type(models, list, "models is a list")
check_length(models, 2, "models has 2 entries")
check_contains(models, 'model-a-v1', "models contains model-a-v1")
check_contains(models, 'model-b-v1', "models contains model-b-v1")
```
