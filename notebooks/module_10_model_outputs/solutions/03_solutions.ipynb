# Module 10 — Notebook 3 Solutions: Heuristics and Patterns

```
import json
import sys
import pandas as pd
from pathlib import Path

sys.path.insert(0, "../../../")
from src.checks import check_equal, check_type, check_length

data_path = Path("../../../data/synthetic/model_outputs.json")
with open(data_path) as f:
    outputs = json.load(f)

df = pd.DataFrame(outputs)
df['response_len'] = df['response'].str.len()
print(f"Loaded {len(df)} records")
```

## Exercise 1 — Solution: Heuristic Flag — "Sure!" Responses

```
df['heuristic_flag'] = df['response'].str.contains('Sure!', na=False)
heuristic_count = int(df['heuristic_flag'].sum())

print(f"Responses matching 'Sure!' heuristic: {heuristic_count}")
df[df['heuristic_flag']][['id', 'model', 'response', 'flagged']]
```

```
# Check Exercise 1
check_type(heuristic_count, int, "heuristic_count is an int")
check_equal(heuristic_count, 2, "heuristic_count is 2")
```

## Exercise 2 — Solution: Filter Short Responses

```
short_responses = df[df['response_len'] < 10]
short_count = len(short_responses)

print(f"Short responses (< 10 chars): {short_count}")
short_responses[['id', 'model', 'response', 'response_len', 'flagged']]
```

```
# Check Exercise 2
check_type(short_count, int, "short_count is an int")
check_equal(short_count, 2, "short_count is 2")
```

## Exercise 3 — Solution: Flagged Outputs Sorted by Response Length

```
flagged_outputs = df[df['flagged'] == True].sort_values('response_len')
shortest_flagged_id = flagged_outputs.iloc[0]['id']

print(f"Shortest flagged output: {shortest_flagged_id}")
flagged_outputs[['id', 'model', 'response', 'response_len', 'category']].head(5)
```

```
# Check Exercise 3
check_type(shortest_flagged_id, str, "shortest_flagged_id is a str")
check_equal(shortest_flagged_id, 'out_015', "shortest flagged output is out_015")
```
