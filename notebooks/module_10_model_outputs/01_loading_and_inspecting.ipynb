## Concept: Batch Analysis of Model Outputs

When you run a model across many prompts, you get a **batch** of outputs. Each output is typically stored as a dict (or JSON object) with several fields. In our dataset, every record has:

| Field | Type | Description |
|---|---|---|
| `id` | str | Unique identifier for this output |
| `model` | str | Which model produced this response |
| `prompt` | str | The input sent to the model |
| `response` | str | The model's reply |
| `flagged` | bool | Whether this output was marked as problematic |
| `category` | str or null | The type of problem (if flagged) |

To analyze a batch, you typically:
1. Load the JSON into a Python list of dicts
2. Inspect one record to understand the structure
3. Compute aggregate statistics across all records

## Worked Example: Counting and Summarizing

Here's how you'd count flagged outputs and compute a flag rate:

```
# Count total outputs
total = len(outputs)
print(f"Total outputs: {total}")

# Count flagged outputs using a list comprehension
example_flagged = sum(1 for r in outputs if r['flagged'])
print(f"Flagged outputs: {example_flagged}")

# Compute flag rate
example_flag_rate = round(example_flagged / total, 4)
print(f"Flag rate: {example_flag_rate}")

# Find unique models
example_models = sorted(set(r['model'] for r in outputs))
print(f"Models: {example_models}")
```
