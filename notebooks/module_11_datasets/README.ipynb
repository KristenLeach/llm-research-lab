# Module 11 — Prompt/Output Datasets

Evaluation datasets are the core artifact of AI research.
This module covers how to create, structure, version, and manage datasets of
prompts, expected behaviors, and model responses.

---

## Learning Objectives

By the end of this module, you will be able to:

- Understand how evaluation datasets are structured (prompts, metadata, expected outputs)
- Create an annotation schema for a classification or rating task
- Build a small dataset with prompts, expected behaviors, and model responses
- Version datasets so you can track changes over time
- Use sampling strategies: random, stratified, adversarial
- Convert between data formats (JSON, CSV, JSONL)
- Understand the difference between a benchmark and an evaluation dataset

---

## What You'll Build

**Mini-project:** Create a small evaluation dataset with:
- 20-30 prompts across 3 task categories
- An annotation schema (what fields does each example have?)
- Expected behaviors for each prompt
- A sampling strategy for selecting diverse examples
- Save in JSONL format (one JSON object per line — the standard for eval datasets)
- Write a dataset card (short description of the dataset)

## Key Concepts

- **Annotation schema**: the definition of what fields each example in the dataset contains
- **JSONL**: JSON Lines format — one JSON object per line, easy to stream and process
- **Dataset card**: documentation describing the dataset's purpose, contents, and limitations
- **Stratified sampling**: ensuring all subgroups are represented in proportion
- **Adversarial examples**: inputs designed to probe model weaknesses
- **Dataset versioning**: using filenames or git to track changes to datasets over time
- **Expected behavior**: what the model should do for a given prompt (vs. what it actually does)
